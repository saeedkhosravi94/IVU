```python
import torch
import detectron2
from detectron2.config import get_cfg
from detectron2.engine import DefaultPredictor
from detectron2.evaluation import COCOEvaluator, inference_on_dataset
from detectron2.data import build_detection_test_loader, DatasetCatalog, MetadataCatalog
from detectron2.utils.visualizer import Visualizer
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import confusion_matrix
import seaborn as sns
import cv2

from detectron2.data.datasets import register_coco_instances
data_dir = '/content/drive/MyDrive/Image and Video Understanding/MRCNN/dataset'


register_coco_instances("taco_train", {}, f"{data_dir}/annotations_train.json", f"{data_dir}/images")
register_coco_instances("taco_test", {}, f"{data_dir}/annotations_test.json", f"{data_dir}/images")
register_coco_instances("taco_val", {}, f"{data_dir}/annotations_val.json", f"{data_dir}/images")


def evaluate_and_plot_confusion_matrix(weights_path, config_path, test_dataset_name):
    """
    Evaluates a trained Mask2Former model on a test dataset and plots a confusion matrix.

    Args:
        weights_path (str): Path to the trained model weights (.pth file).
        config_path (str): Path to the model configuration file (.yaml).
        test_dataset_name (str): The name of the registered test dataset.
    """
    cfg = get_cfg()
    try:
        cfg.merge_from_file(config_path)
    except KeyError as e:
        if "'MODEL.RESNETS.STEM_TYPE'" in str(e):
            print("Ignoring 'MODEL.RESNETS.STEM_TYPE' from config file due to version mismatch.")

            pass

    if hasattr(cfg.MODEL.RESNETS, 'STEM_TYPE'):
         try:
             del cfg.MODEL.RESNETS.STEM_TYPE
             print("Successfully removed MODEL.RESNETS.STEM_TYPE.")
         except AttributeError:
             print("MODEL.RESNETS does not have STEM_TYPE attribute after merge attempt.")


    cfg.MODEL.WEIGHTS = weights_path
    cfg.MODEL.RETINANET.SCORE_THRESH_TEST = 0.5

    predictor = DefaultPredictor(cfg)

    evaluator = COCOEvaluator(test_dataset_name, output_dir="./output")

    val_loader = build_detection_test_loader(cfg, test_dataset_name)

    print("Running inference and collecting data for confusion matrix...")

    true_labels = []
    pred_labels = []

    dataset_dicts = DatasetCatalog.get(test_dataset_name)
    metadata = MetadataCatalog.get(test_dataset_name)
    class_names = metadata.thing_classes

    for i, data in enumerate(dataset_dicts):
        img = cv2.imread(data["file_name"])
        outputs = predictor(img)


        for annotation in data.get("annotations", []):
            if "category_id" in annotation:
                true_labels.append(annotation["category_id"])


        if "instances" in outputs:
            instances = outputs["instances"].to("cpu")

            for j in range(len(instances)):
                pred_labels.append(instances.pred_classes[j].item())

    print("Data collection for confusion matrix completed.")



    if true_labels and pred_labels:
        all_labels = list(set(true_labels + pred_labels))
        all_labels.sort()

        label_to_index = {label: i for i, label in enumerate(all_labels)}
        mapped_true_labels = [label_to_index[label] for label in true_labels]
        mapped_pred_labels = [label_to_index[label] for label in pred_labels]


        cm = confusion_matrix(mapped_true_labels, mapped_pred_labels)
        plt.figure(figsize=(10, 8))
        display_class_names = [class_names[label] for label in all_labels]
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=display_class_names, yticklabels=display_class_names)
        plt.xlabel('Predicted')
        plt.ylabel('True')
        plt.title('Confusion Matrix')
        plt.show()
    else:
        print("Could not generate confusion matrix: Data not collected or no instances found.")



weights_path = "/content/drive/MyDrive/Image and Video Understanding/M2FORMER/output/mask2former_r101/model_final.pth"
config_path = "/content/drive/MyDrive/Image and Video Understanding/M2FORMER/Mask2Former/configs/coco/instance-segmentation/config.yaml"
test_dataset_name = "taco_test"

evaluate_and_plot_confusion_matrix(weights_path, config_path, test_dataset_name)
```
