```python
def evaluate_yolo_box_seg(model_path, data_yaml):

    model = YOLO(model_path)
    results = model.val(data=data_yaml, split='val', verbose=False)

    metrics = {}

    # Bounding box metrics
    if results.box is not None:
        metrics["box_precision"] = results.box.p.mean().item()
        metrics["box_recall"] = results.box.r.mean().item()
        metrics["box_map50"] = results.box.map50.item()

    # Segmentation metrics
    if results.seg is not None:
        metrics["seg_precision"] = results.seg.p.mean().item()
        metrics["seg_recall"] = results.seg.r.mean().item()
        metrics["seg_map50"] = results.seg.map50.item()

    return metrics
```

```python
import torch
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
import yaml
import os

def visualize_yolo_prediction(weights_path, data_yaml_path, image_path):
    # === Load model ===
    model = YOLO(weights_path)  # works for both YOLOv5/YOLOv8 (Ultralytics version)

    # === Load class names from data.yaml ===
    with open(data_yaml_path, 'r') as f:
        data_yaml = yaml.safe_load(f)
    class_names = data_yaml['names'] if 'names' in data_yaml else []

    # === Read image ===
    image = cv2.imread(image_path)
    if image is None:
        raise ValueError(f"Could not read image: {image_path}")
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # === Run inference ===
    results = model(image_path)[0]  # single image, get first result

    # === Plot predictions ===
    for box in results.boxes:
        cls_id = int(box.cls)
        conf = float(box.conf)
        x1, y1, x2, y2 = map(int, box.xyxy[0])  # box.xyxy is (N, 4) tensor

        label = f"{class_names[cls_id]} {conf:.2f}" if class_names else f"Class {cls_id} {conf:.2f}"
        cv2.rectangle(image_rgb, (x1, y1), (x2, y2), color=(0, 255, 0), thickness=2)
        cv2.putText(image_rgb, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX,
                    0.6, (0, 255, 0), 2)

    # === Display result ===
    plt.figure(figsize=(10, 10))
    plt.imshow(image_rgb)
    plt.axis("off")
    plt.title("YOLO Predictions")
    plt.show()
```

```python
weights_path = "/content/drive/MyDrive/Image and Video Understanding/YOLO/results/train_200_960_16/weights/best.pt"
data_yaml_path = "/content/drive/MyDrive/Image and Video Understanding/YOLO/dataset/data.yaml"
image_path = "/content/drive/MyDrive/Image and Video Understanding/YOLO/dataset/test/images/401.jpg"

visualize_yolo_prediction(weights_path, data_yaml_path, image_path)
```

```python
import os
import cv2
import torch
import yaml
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

def visualize_yolo_predictions_with_masks(weights_path, data_yaml_path, image_folder):
    # Load YOLO segmentation model
    model = YOLO(weights_path)

    # Load class names
    with open(data_yaml_path, 'r') as f:
        data_yaml = yaml.safe_load(f)
    class_names = data_yaml.get('names', [])

    # Supported image extensions
    supported_exts = ('.jpg', '.jpeg', '.png')
    image_files = [f for f in os.listdir(image_folder) if f.lower().endswith(supported_exts)]

    if not image_files:
        print("No images found in the folder.")
        return

    for img_name in image_files:
        image_path = os.path.join(image_folder, img_name)
        image_bgr = cv2.imread(image_path)
        if image_bgr is None:
            print(f"⚠️ Could not read image: {image_path}")
            continue

        image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
        img_h, img_w = image_rgb.shape[:2]

        # Run inference
        results = model(image_path)[0]

        # Overlay masks if available
        if hasattr(results, "masks") and results.masks is not None:
            for idx, mask in enumerate(results.masks.data):
                # mask is usually smaller than original image - resize it
                mask_np = mask.cpu().numpy().astype(np.uint8)  # float mask 0..1 -> 0 or 1 after threshold
                mask_resized = cv2.resize(mask_np, (img_w, img_h), interpolation=cv2.INTER_NEAREST)

                # Create a color mask overlay (green here)
                colored_mask = np.zeros_like(image_rgb, dtype=np.uint8)
                colored_mask[mask_resized > 0.5] = (0, 255, 0)  # green

                # Blend original image and mask overlay
                image_rgb = cv2.addWeighted(image_rgb, 1.0, colored_mask, 0.5, 0)

                # Optionally draw contours around mask
                contours, _ = cv2.findContours((mask_resized > 0.5).astype(np.uint8), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
                cv2.drawContours(image_rgb, contours, -1, (0, 255, 0), 2)

        # Draw bounding boxes and labels
        for box in results.boxes:
            cls_id = int(box.cls)
            conf = float(box.conf)
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            label = f"{class_names[cls_id]} {conf:.2f}" if class_names else f"Class {cls_id} {conf:.2f}"

            cv2.rectangle(image_rgb, (x1, y1), (x2, y2), (255, 0, 0), 2)
            cv2.putText(image_rgb, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2)

        # Show result
        plt.figure(figsize=(10, 8))
        plt.imshow(image_rgb)
        plt.title(f"Predictions: {img_name}")
        plt.axis("off")
        plt.show()
```

```python
weights_path = "/content/drive/MyDrive/Image and Video Understanding/YOLO/results/train_200_960_16/weights/best.pt"
data_yaml_path = "/content/drive/MyDrive/Image and Video Understanding/YOLO/dataset/data.yaml"
image_folder = "/content/drive/MyDrive/Image and Video Understanding/YOLO/dataset/test/images"

visualize_yolo_predictions_with_masks(weights_path, data_yaml_path, image_folder)
```
